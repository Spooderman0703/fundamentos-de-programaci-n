**Extra 1: Control de aforo con break y continue (while)**

Enunciado: Crea un programa que procese visitantes del museo con un ciclo while. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa continue). Si el total acumulado alcanza o supera $100, detén el ciclo con break y muestra el total.



```python
total_acumulado = 0
numero_boleto = 1

while True:
    edad = int(input(f"Ingrese la edad del visitante {numero_boleto}: "))
    if edad < 3:
        precio_boleto = 0
    elif 3 <= edad <= 17:
        precio_boleto = 30
    else:
        precio_boleto = 45
    if precio_boleto == 0:
        print("Menor de 3 años, sin cargo")
        numero_boleto += 1
        continue

    total_acumulado += precio_boleto
    print(f"Costo del boleto {numero_boleto}: ${precio_boleto}")
    if total_acumulado >= 100:
        print(f"Aforo alcanzado, se detiene el registro.\nTotal acumulado: ${total_acumulado}")
        break

    numero_boleto += 1
```

Se utilizaron estructuras de repetición como: **While**, para generar un bucle y procesar un número determinado de visitantes, definido por la petición del código. También se utilizó **Break** para detener el código una vez que el total acumulado llegara o se sobrepasara de "$100".**Continue** fue el control de flujo que notifica que no hay cargo que se deba de hacer, de acuerdo a la edad del visitante, y que genera un salto al resto de código para que vuelva al inicio, sin acumular nada.

También se usaron estructuras de decisión como: **if** que empieza con la estructura de evaluación si es que se cumple alguna de las condiciones de edad, para esto mismo sirve **elif** y **else**. 


**Extra 2: Estadística de visitantes con for**

Enunciado: Crea un programa que pida el número total de visitantes y, con un ciclo for, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales.

```python
numero_visitantes = int(input("¿Cuál es el número de visitantes?"))
adultos_visitantes = 0
total=0

for i in range(1, numero_visitantes+1):
    edad_visitante = float(input("¿Qué edad tiene el visitante?: "))
    print(f"Edad del visitante {i} : {edad_visitante}")

    total = float(total + edad_visitante)

    if edad_visitante >= 18:
        adultos_visitantes += 1
        

    

total = total / i


print(f"Número de visitantes : {i}")
print(f"\nPromedio de edad: {total}")
print(f"Adultos: {adultos_visitantes}.0")



```

Se volvió a usar una estructura de repetición, en este caso, fue **for** el que inicializa este sistema que se complementa con **in** y **range** para determinar cuántas veces se repetirá este ciclo. El número que define estos ciclos es el valor asignado de la primer petición.

También se usaron estructuras de decisión como **if** que aplica una condicional simple para filtrar y contabilizar únicamente a la población adulta.

**Extra 4: Pirámide de asteriscos con for**

Enunciado: Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos for anidados y el parámetro end de print.

```python
altura = int(input("¿Cuántos niveles de altura deseas para la pirámide?"))

print(f"Altura (niveles) de la pirámide: {altura}")

for i in range(1, altura +1):
    for j in range(1, i + 1):
        print("*", end="")
    print()

```

Este código es bastnate simple, por lo que solamente se usaron estructuras de repetición: **for** fue el inicializador del ciclo de repetición, y que trabajó en conjunto con la variable "altura" para definir cuántos niveles tendrá la pirámide. 

Se usó también otro **for** debajo del primero establecido, con la función de añadir asteriscos con cada nuevo salto de nivel. La propiedad *end=""* evita que Python aplique un salto de línea por cada asterisco, obligando la impresión en un único renglón horizontal.
