```python
def mostrar_tabla(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

def suma_fila(fila):
    suma_total = 0
    for e in fila:
        suma_total += e
    return suma_total

matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

mostrar_tabla(matriz)

print(f"Suma de la fila 0: {suma_fila(matriz[0])}")
print(f"Suma de la fila 1: {suma_fila(matriz[1])}")
print(f"Suma de la fila 2: {suma_fila(matriz[2])}")

```
