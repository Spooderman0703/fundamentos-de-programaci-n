**Extra 1: Sistema de calificaciones con tuplas**

Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`. Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor a menor** y muestra la suma de todos los elementos usando una función que retorne el valor.

```python
def suma_elementos(lista_calif):
    total = 0
    for i in lista_calif:
        total += i
    return total


calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)


print(f"La tercera calificación de la tupla es: {calificaciones[2]}\n")

print("Calificaciones nuevas:")
calif1 = float(input("Ingresa una nueva calificación: "))
print(f"La calificación ingresada fue: {calif1}")
calif2 = float(input("Ingresa otra califación: "))
print(f"La calificación ingresada fue: {calif2}\n")

nueva_lista = calificaciones + (calif1, calif2)

lista_calif = list(nueva_lista)

lista_calif.sort(reverse=True)

print(f"Lista con calificaciones nuevas integradas: {lista_calif}")

print(f"La suma de todas las calificaciones de la nueva tupla es: {suma_elementos(lista_calif)}")
```

**Extra 2: Agenda de contactos con búsqueda**

Crea un diccionario `agenda` con tres contactos iniciales (`"Ana": "555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto (nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos los nombres con `keys()` en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto.


```python
def buscar_telefono(agenda):
    busq_nombre = input("Ingresa el nombre del contacto para buscar su número telefónico: ")

    if busq_nombre in agenda:
        telefono = agenda[busq_nombre]
        return f'El número telefónico de "{busq_nombre}" es: "{telefono}".'
    else:
        return f'Contacto no encontrado.'



agenda = {
    "Ana" : 5550101,
    "Luis" : 5550102,
    "Mía" : 5550103
}

nombre = input("Ingresa el nombre del nuevo contacto: ")
num_telefono = int(input("Ingresa su número telefónico: "))

agenda[nombre] = num_telefono

print("Los nombres de la agenda son: ", end="")
for nombre in agenda.keys():
    print(nombre, end= ", ")

buscar_telefono(agenda)
```

**Desafío Extra 2: Contador de votos con diccionario**

Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con `items()` el color ganador junto con su total.

```python
candidatos = {
    "rojo": 0,
    "azul": 0,
    "verde": 0
}

cantidad = 5
conteo = 1


while conteo <= cantidad:    
    voto = input(f"Ingresa el voto {conteo} (rojo, azul, verde): ")
    if voto.lower() in candidatos:
        candidatos[voto.lower()] += 1
        conteo += 1
    else:
        print("El voto no es válido")

print("\nResultados:")
for k, v in candidatos.items():
    print(f"Opción {k}, votos {v}")

if candidatos["rojo"] > candidatos["azul"] and candidatos["rojo"] > candidatos["verde"]:
    print(f"El ganador de las votaciones fue el color rojo, con {candidatos["rojo"]} votos.")
elif candidatos["azul"] > candidatos["rojo"] and candidatos["azul"] > candidatos["verde"]:
    print(f"El ganador de las votaciones fue el color azul, con {candidatos['azul']} votos.")
elif candidatos["verde"] > candidatos["azul"] and candidatos["verde"] > candidatos["rojo"]:
    print(f"El ganador de las votaciones fue el color verde, con {candidatos['verde']} votos.")
else: 
    print("Hubo empate de votos en el primer lugar.")


    
```

**Desafío Extra 3: Validación robusta de entrada**

Escribe un programa que pida al usuario dos números enteros dentro de un bloque `try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo `while`. Luego pide la división del primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
while True:
    try:
        a = int(input("Ingresa el dividendo: "))
        b = int(input("Ingresa el divisor: "))
        print(f"El resultado de la división es: {a / b}")
        break
    except ValueError:
        print("Ingresaste un valor incorrecto, ingresa un valor entero.")
    except ZeroDivisionError:
        print("No es posible dividir un número entre cero, ingresa nuevos números.")




```

**Desafío Extra 4: Analizador de textos**

Crea un programa que reciba una frase, imprima su longitud con `len()`, la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
def contar_palabras(mensaje):
    return len(mensaje.split())

mensaje = input("Ingresa la frase que desees: ") # "amo python"
print(f"La longitud de tu frase ('{mensaje}') es de: {len(mensaje)} letras.")

nuevo_mensaje = (mensaje.replace("python", "programación"))

print(f"Tu nueva frase es: {nuevo_mensaje.upper()}")


total = contar_palabras(mensaje)
print(f"El total de palabras de tu frase final es: {total}.")
```
