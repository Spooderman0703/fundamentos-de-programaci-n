**Ejercicio 1: Área de un triángulo**

Escribe un programa en Python que aplique el modelo Entrada-Proceso-Salida para calcular el área de un triángulo (area = (base * altura) / 2), usando input(), float() y print().

**Ejercicio 10: Suma acumulada con while**

Pide números mientras sean distintos de 0 (el 0 termina el ciclo y no se suma).

```python
lista_num = []
total = 0

while True:
    num = int(input("Ingresa un número para continuar con la suma (Al ingresar '0' se termina el programa): "))
    lista_num.append(num)

    if num == 0:
        for i in lista_num:
            total += i
        lista_num.remove(0)
        print(f"La lista completa de números es: {lista_num}\n")
        print(f"La suma de todos los números es: {total}")

        break

    
```

**Ejercicio 12: Suma de pares del 1 al 20 con for y continue**

Salida: La suma de los números pares del 1 al 20 es: 110

```python
num_par = 0

for i in range(1, 21):
    if i % 2 == 0:
        num_par += i
        print(f"El número {i} es un número par.")
    else:
        continue
    
print(f"\nLa suma de pares (1 - 20) es: {num_par}")
```

**Ejercicio 19: Tupla: acceso y suma con función**

Entrada: (4, 8, 15, 16, 23, 42) → Salida:

```python

def suma_tupla(tupla):
    total = 0
    for i in tupla:
        total += i    
    return total


tupla = (4, 8, 15, 16, 23, 42)

print(f"La tupla completa es: {tupla}\n")
print(f"El tercer elemento de la tupla es: {tupla[2]}")
print(f"la suma de todos los elementos de la tupla es: {suma_tupla(tupla)}")

```

**Ejercicio 21: Diccionario de calificaciones y promedios**

Entrada: {"Ana": [8, 9, 10], "Beto": [7, 7, 8]} → Salida:

```python
def promedio(diccionario, nombre):
    total = 0
    num_calif = 0
    for i in diccionario.get(nombre):
        total += i
        num_calif += 1
        prom = total / num_calif
    return print(f"El promedio de {nombre} es: {prom}")

diccionario = {
    "Ana" : [8, 9, 10],
    "Beto" : [7, 7, 8], 
    "Arturo" : [6, 8, 5]
}


promedio(diccionario, "Ana")
promedio(diccionario, "Beto")


```

**Ejercicio 24: Contar palabras de un string**

Entrada: "Python es un lenguaje de programacion" → Salida: El mensaje tiene 6 palabras.

```python
def contar_palabras(mensaje):
    return len(mensaje.split())

mensaje = "Python es un lenguaje de programación"

total = contar_palabras(mensaje)

print(f"El mensaje tiene: {total} palabras.")
```
