**Ejercicio 1: Área de un triángulo**

Escribe un programa en Python que aplique el modelo Entrada-Proceso-Salida para calcular el área de un triángulo (area = (base * altura) / 2), usando input(), float() y print(). Entrada: base = 8, altura = 5 Salida: El área del triángulo es: 20.0

```python
base = float(input("Ingresa el valor de la base: "))
altura = float(input("Ingresa el valor de la altura: "))

area = ((base * altura) / 2)

print(f"El área del triángulo es: {area}")
```

**Ejercicio 2: Conversión de Celsius a Fahrenheit**

Escribe un programa en Python que pida una temperatura en grados Celsius y la convierta a Fahrenheit (F = C * 9/5 + 32). Entrada: celsius = 30 Salida: 30 grados Celsius equivalen a 86.0 grados Fahrenheit

```python
temp_celsius = float(input("Ingresa una temperatura en grados Celsius para hacer su converisón: "))

temp_fahrenheit = (temp_celsius * 9 / 5 + 32)

print(f"{temp_celsius} grados Celsius equivalen a {temp_fahrenheit} grados Fahrenheit.")
```

**Ejercicio 3: Interés simple (antes en PSeInt)**

Escribe un programa en Python que pida capital, tasa y tiempo, y calcule el interés simple (interes = capital * (tasa / 100) * tiempo). Entrada: capital = 2000, tasa = 6, tiempo = 2 Salida: El interés simple es: 240.0

```python
capital = float(input("Ingresa el número de capital: "))
tasa = float(input("Ingresa el número de tasa: "))
tiempo = float(input("Ingresa el número de tiempo: "))

int_simple = (capital * (tasa / 100) * tiempo)

print(f"El interés simple es: {int_simple}")
```

**Ejercicio 5: Casting tras input() (antes de análisis teórico)**

Escribe un programa en Python que pida un número con input(), demuestre que devuelve str (type()), intente x + 1 dentro de try-except capturando TypeError, y finalmente haga int(x) para sumar y obtener el resultado correcto. Entrada: "25" Salida: x es tipo str, x + 1 genera TypeError, int(x) + 1 = 26

```python
num = input("Ingresa un número (como cadena): ")
print(f"El número ingresado es de tipo: {type(num)}. \n")

try:
    resultado = num + 1
except TypeError:
    print(f"No es posible sumar {num} + 1, ya que {num} está almacenado como tipo 'str'. \n")

num_int = int(num)
print(f"{num} fue convertido a tipo: {type(num_int)}; La suma {num_int} + 1 = {num_int + 1}")
```

**Ejercicio 6: Conversión de horas a segundos**

Crea un programa que pida una cantidad de horas (float) y la convierta a segundos (1 hora = 3600 segundos), mostrando el resultado formateado. Entrada: 3 Salida: 3 horas equivalen a 10800 segundos

**Ejercicio 7: Descuento del 15% con f-strings**

Crea un programa que pida el precio de un artículo y aplique un 15% de descuento, mostrando el monto del descuento y el precio final con dos decimales usando f-strings. 

```python
precio = int(input("Ingresa el precio de un artículo: "))

print(f"El precio original fue: ${precio}, se aplicó un 15% de descuento: ${precio * 0.15}, y su precio final es de: ${precio - (precio * 0.15)}")
```
